```python
# handling date of journey
df['Date_of_Journey'] = pd.to_datetime(df['Date_of_Journey'], format='%d/%m/%Y')
df['Day'] = df['Date_of_Journey'].dt.day.astype(int)
df['Month'] = df['Date_of_Journey'].dt.month.astype(int)
df['Year'] = df['Date_of_Journey'].dt.year.astype(int)
df.drop('Date_of_Journey', axis=1, inplace=True)
```

```python
# handling arrival time

df['Arrival_Time'] = pd.to_datetime(df['Arrival_Time'].apply(lambda x: x.split(' ')[0]), format='%H:%M')
df['Arrival_Hour'] = df['Arrival_Time'].dt.hour.astype(int)
df['Arrival_Minute'] = df['Arrival_Time'].dt.minute.astype(int)
df.drop('Arrival_Time', axis=1, inplace=True)
```

```python
# handle departure time
df['Dep_Time'] = pd.to_datetime(df['Dep_Time'], format='%H:%M')
df['Dep_Hour'] = df['Dep_Time'].dt.hour.astype(int)
df['Dep_Minute'] = df['Dep_Time'].dt.minute.astype(int)
df.drop('Dep_Time', axis=1, inplace=True)
```

```python
df.head()
```

```python
# Convert duration to minutes
def convert_duration_to_minutes(duration):
    duration = duration.strip()
    if 'h' in duration and 'm' in duration:
        hours, minutes = duration.split('h')
        minutes = minutes.replace('m', '').strip()
        return int(hours) * 60 + int(minutes)
    elif 'h' in duration:
        hours = duration.replace('h', '').strip()
        return int(hours) * 60
    elif 'm' in duration:
        minutes = duration.replace('m', '').strip()
        return int(minutes)
    else:
        return 0

df['Duration_in_Minutes'] = df['Duration'].apply(convert_duration_to_minutes).astype(int)
df.drop('Duration', axis=1, inplace=True)
```
