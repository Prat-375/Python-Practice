```python
# Data Encoding
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a dataframe

df = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'red', 'blue']
})
```

```python
df.head()
```

```python
# Create an instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
# perform fit and transform
encoded_data = encoder.fit_transform(df[['color']]).toarray()
```

```python
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out())
```

```python
encoded_df.head()
```

```python
# for new data
encoder.transform([['red']]).toarray()
```

```python
pd.concat([df, encoded_df], axis=1)
```

```python
# Ordinal Encoding

from sklearn.preprocessing import OrdinalEncoder
ordinal_encoder = OrdinalEncoder()

df_ordinal = pd.DataFrame({
    'size': ['small', 'medium', 'large', 'medium', 'small']
})
```

```python
# Create an instance of OrdinalEncoder
ordinal_encoder = OrdinalEncoder(categories=[['small', 'medium', 'large']])

# perform fit and transform
ordinal_encoded_data = ordinal_encoder.fit_transform(df_ordinal[['size']])

ordinal_encoded_df = pd.DataFrame(ordinal_encoded_data, columns=ordinal_encoder.get_feature_names_out())
ordinal_encoded_df.head()
```

```python
# Target guided ordinal encoding
df = pd.DataFrame({
    'city': ['A', 'B', 'C', 'D', 'D', 'C', 'A', 'A'],
    'price': [100, 150, 200, 250, 300, 350, 400, 450]
})

target_mean = df.groupby('city')['price'].mean().sort_values()
```

```python
df['city_encoded'] = df['city'].map(target_mean)
```

```python
df[['price', 'city_encoded']]
```

```python
import seaborn as sns
tips_data = sns.load_dataset('tips')
```

```python
tips_data.head()
```

```python
tips_data['time'].unique()
```

```python
tips_target_mean = tips_data.groupby('time')['total_bill'].mean()
```

```python
tips_data['time_encoded'] = tips_data['time'].map(tips_target_mean)
```

```python
tips_data[['time', 'time_encoded']].drop_duplicates()
```

```python
tips_data['time_encoded'].unique()

# Shows dinner is more expensive than lunch
# This is a very simple encoding technique but it can be very powerful. 
# It can capture the relationship between the categorical variable and the target variable. 
# However, it can also lead to overfitting if there are too many categories or if the relationship is not strong enough.
```

```python
lunch_count = (tips_data['time'] == 'Lunch').sum()
dinner_count = (tips_data['time'] == 'Dinner').sum()

print(f'Lunch count: {lunch_count}')
print(f'Dinner count: {dinner_count}')
```

```python
import seaborn as sns
tips_data = sns.load_dataset('tips')
```

```python
tips_majority = tips_data[tips_data['time'] == 'Dinner']
tips_minority = tips_data[tips_data['time'] == 'Lunch']

tips_minority_upsampled = tips_minority.sample(n=len(tips_majority), replace=True, random_state=42)
tips_upsampled = pd.concat([tips_majority, tips_minority_upsampled])

tips_upsampled['time'].value_counts()
```

```python
tips_target_mean = tips_upsampled.groupby('time')['total_bill'].mean()
```

```python
tips_upsampled['time_encoded'] = tips_upsampled['time'].map(tips_target_mean)
```

```python
tips_upsampled[['time', 'time_encoded']].drop_duplicates()

# This is a simple example of how to handle imbalanced data using upsampling. 
# We have two classes: Lunch and Dinner. Dinner is the majority class and Lunch is the minority class. We upsample the minority class by randomly sampling with replacement until we have the same number of samples in both classes.
#  Then we perform target guided ordinal encoding on the upsampled data.

# We find that the mean total bill for Lunch is 17.28 and for Dinner is 20.80.
#  This shows that Dinner is more expensive than Lunch on average.
```
