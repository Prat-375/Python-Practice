```python
## Handling Imbalanced Dataset
import numpy as np
import pandas as pd

# Set the random seed for reproducibility
np.random.seed(123) 
# Why 123? It's a common practice to use a specific number as a seed for reproducibility. 
# The choice of 123 is arbitrary, but it ensures that the random processes in the code will yield the 
# same results each time it is run, which is important for debugging and comparing results.

# Create a dataframe with 1000 samples and 2 features
n_samples = 1000
class_0_ratio = 0.9  # 90% of samples belong to class 0
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0, n_class_1
```

```python
# Create dataframe with imbalanced classes
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=1, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=1, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0, class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df['target'].value_counts()
```

```python
## Upsampling the minority class
df_minority= df[df['target']==1]
df_majority= df[df['target']==0]
```

```python
from sklearn.utils import resample 
# resample is a function from the sklearn.utils module that allows you to perform random sampling of your data.

df_minority_upsampled = resample(df_minority,
         replace=True,  # Sample with replacement
         n_samples=len(df_majority),  # Match the number of samples in the majority class
         random_state=42)  # Set a random seed for reproducibility
```

```python
df_minority_upsampled.shape
```

```python
df_minority_upsampled.head()
```

```python
df_upsampled=pd.concat([df_majority, df_minority_upsampled])
```

```python
df_upsampled['target'].value_counts()
```
